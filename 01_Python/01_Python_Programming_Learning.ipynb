```python
nums = [1,2,10,5,7,22,18]
largest = nums[0]
slargest = -1

for i in range(len(nums)):
  if largest < nums[i] :
    slargest = largest
    largest = nums[i]
  elif slargest < nums[i] and nums[i] != largest :
    slargest = nums[i]
print(slargest)
```

```python
class Student :
  def __init__(self,name,age):
    self.name = name
    self.age = age

s1 =  Student("spidey",21)
print(s1.name)
print(s1.age)
```

```python
class Student :

  def __init__(self,name):
    self.name = name

  def display(self):
    print(self.name)

s1 = Student("Sam")
s1.display()
```

```python
class Animal:
  def eat(self):
    print("Eating")

class Dog(Animal) :

  def bark(self):
    print("Barking")

d = Dog()
d.eat()
d.bark()
```

```python
# Exception handling
num1 = int(input("Enter first number : "))
num2 = int(input("Enter second number : "))

try:
    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError:
    print("Cannot divide by zero!")

except ValueError:
    print("Please enter valid integers.")
```
