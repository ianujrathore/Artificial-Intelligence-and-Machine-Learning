```python
# WAP a program To Find Factorial of The Number
num = int(input('Enter the number : '))

if (num < 0) :
   print('Factorial for negative integer does not exist!')
   exit()
else :
  fact = 1;
  for i in range (1,num + 1) :
    fact *= i;
  print(fact)
```

```python
# WAP to check if the number is prime or not prime
num = int(input('Enter a number : '))

if num <= 1:
    print('Not Prime')

else:
    for i in range(2, num):

        if num % i == 0:
            print('Not Prime')
            break

    else:
        print('Prime')
```

```python
# Fibonacci using recursion

def fib(num):

    if num == 1:
        return [0]

    elif num == 2:
        return [0, 1]

    else:
        series = fib(num - 1)
        series.append(series[-1] + series[-2])
        return series


num = int(input("Enter number: "))

print(fib(num))
```

```python
# Power of a number using recursion

def power(base, exponent):

    # Base condition
    if exponent == 0:
        return 1

    # Recursive call
    return base * power(base, exponent - 1)


base = int(input("Enter base number: "))
exponent = int(input("Enter exponent: "))

result = power(base, exponent)

print("Result:", result)
```
