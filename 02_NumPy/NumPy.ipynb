## **Learn about NumPy**

```python
import numpy as np
import time
```

### Creating Array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("1D Array : ", arr_1d)
```

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D Array : ", arr_2d)
```

### List vs NumPy Array

```python
py_list = [1,2,3]
print("Python List multiplication ", py_list * 2)

py_array = np.array([1,2,3])
print("Python Array multiplication ", py_array * 2)
```

```python
start = time.time()
py_list = [i*2 for i in range(100000)]
print("\n List operation time : " ,time.time() - start)

start = time.time()
py_array = np.arange(100000) * 2
print("\n Numpy operation time : ", time.time()- start)
```

### Creating array from scratch

```python
zeros = np.zeros((3,4))
print("Zero Matrix : \n", zeros)

ones = np.ones((3,4))
print("\n Ones Matrix : \n", ones)

empty = np.empty((3,4))
print("\n Empty Matrix : \n", empty)

full = np.full((3,4),5)
print("\n Full Matrix : \n", full)

identity = np.eye(3)
print("\n Identity Matrix : \n", identity)

random = np.random.random((3,4))
print("\n Random Matrix : \n", random)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector : ",vector)

matrix = np.array([[1,2,3],[4,5,6]])
print("\nMatrix : ",matrix)

tensor = np.array([[[1,2], [3,4]],[[5,6],[7,8]]])
print("\nTensor : ",tensor)
```

### Numpy / Array Properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print("Shape : ", arr.shape)
print("Size : ", arr.size)
print("Dimension : ", arr.ndim)
print("Data Type : ", arr.dtype)
```

```python
# Reshape array
arr = np.arange(12)
print("Original Array : ", arr)

reshaped = arr.reshape((3,4))
print("\nReshaped Array : ", reshaped)

flattened = reshaped.flatten()
print("\nFlattened Array : ", flattened)

transposed = reshaped.transpose()
print("\nTransposed Array : ", transposed)
```

## **Numpy** **Array** **Operation**

```python
arr = np.array([1,2,3,4,5,6])
print("Basic Slicing : ",arr[2:5])
print("With Step : ", arr[1:5:2])
print("Negative Indexing : ", arr[-3])
```

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print("Speciic Element : ",arr_2d[1,2])
print("Entire Row : ", arr_2d[1])
print("Entire Row : ", arr_2d[:, 1])
```

```python
# Sorting
unsorted = np.array([3,1,4,5,6,3,9,67,43,2,1])
print("Unsorted Array : ", unsorted)

sorted = np.sort(unsorted)
print("Sorted Array : ", sorted)

unsorted_2d = np.array([[3,9,4],[5,8,7]])
print("Unsorted 2D Array : ", unsorted_2d)

sorted_2d = np.sort(unsorted_2d,axis = 0) # Use axis 0 to sort columns and axis 1 row wise
print("Sorted 2D Array : ", sorted_2d)
#
```

### Adding and Removing data

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

print("Concatenation : ", np.concatenate((arr1,arr2)))
```

### **Where**

```python
a = np.array([10,20,30,40,50,60])
print(np.where(a > 25))
print("OR")
print(np.where(a > 25, 1, 0))
```
