## **Normalization & Standardization**

Q49. Perform Min-Max normalization.

Formula:
(x - min) / (max - min)

Q50. Standardize the subject scores.

Formula:
(x - mean) / std

Q51. Verify that your standardized data has
    approximately mean = 0.

Q52. Verify that its standard deviation is
    approximately 1.

```python
subjects = students[:, 2:5]
normalized = (subjects - subjects.min(axis=0)) / (subjects.max(axis=0) - subjects.min(axis=0))
print(normalized)
```

```python
subjects = students[:, 2:5]
standardized = (subjects - subjects.mean(axis=0)) / subjects.std(axis=0)

print(standardized)
```

```python
standardized.mean(axis=0)
```

```python
standardized.std(axis=0)
```

```python
# Create features X and target Y
# X = Math, Physics, Chemistry
# y = Final_Score

X = students[:, 2:5]
Y = students[:, -1]
```

```python
# Shape of X
X.shape
```

```python
# Shape of Y
Y.shape
```

```python
# Mean of X
X.mean(axis=0)
```

```python
# Standardized
X_standardized = (X - X.mean(axis=0)) / X.std(axis=0)
print(X_standardized)
```

```python
# Create random train/test indices
np.random.seed(42)                 # same random result every time
indices = np.random.permutation(len(X))
print(indices)
```

```python
# Split the data into 80% training and 20% testing
split = int(0.8 * len(X))
train_indices = indices[:split]
test_indices = indices[split:]

X_train = X[train_indices]
X_test = X[test_indices]

Y_train = Y[train_indices]
Y_test = Y[test_indices]
```

```python
# Check the shapes of X_train, X_test, y_train and y_test.
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)
```

```python
# Calculate the mean of each training feature.
X_train_mean = X_train.mean(axis=0)
print(X_train_mean)
```

```python
# Use those training statistics to standardize both training and test data.
X_train_standardized = (X_train - X_train_mean) / X_train.std(axis=0)
X_test_standardized = (X_test - X_train_mean) / X_train.std(axis=0)
```
