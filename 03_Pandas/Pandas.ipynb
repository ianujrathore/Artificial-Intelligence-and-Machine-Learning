## Missing Data

```python
# Data and Creating a Dataframe

data = {
    "name": ["Peter", "John", "Sam", "Sara", "Virat", "Rahul", "Anita", "David"],
    "age": [21, 24, None, 27, 23, None, 25, 22],
    "score": [85, None, 78, 95, None, 88, 91, None],
    "city": ["Delhi", "Mumbai", None, "Bangalore", "Mumbai", None, "Pune", "Delhi"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
# Find missing value in columns
df.isna().any()
```

```python
# Remove all the row with null values
# df.dropna()
```

```python
# Filling the missing data
df.fillna(0)
```

```python
# Fill age with mean
df["age"] = df["age"].fillna(df["age"].mean())

# Fill score with mean
df["score"] = df["score"].fillna(df["score"].mean())

# Fill city with mode
df["city"] = df["city"].fillna(df["city"].mode()[0])

print(df)
```

## Merging Joining Concatination

```python
import pandas as pd

data1 = {
    "id": [1, 2, 3, 4],
    "name": ["Peter", "John", "Sam", "Sara"]
}

data2 = {
    "id": [1, 2, 3, 4],
    "score": [85, 92, 78, 95]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

df = pd.merge(df1, df2, on="id")

print(df)
```

```python
data1 = {
    "name": ["Peter", "John"],
    "score": [85, 92]
}

data2 = {
    "name": ["Sam", "Sara"],
    "score": [78, 95]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

df = pd.concat([df1, df2], ignore_index=True)

print(df)
```

```python
df1 = pd.DataFrame({
    "name": ["Peter", "John", "Sam"],
    "score": [85, 92, 78]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Pune"]
}, index=[1, 2, 3])

df = df1.join(df2)

print(df)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Peter", "John", "Sam"]
})

df2 = pd.DataFrame({
    "id": [2, 3, 4],
    "score": [92, 78, 95]
})

df = pd.merge(df1, df2, on="id", how="outer")

print(df)
```

## GroupBy and Aggregation

```python
import pandas as pd

data = {
    "name": ["Peter", "John", "Sam", "Sara", "Virat", "Rahul"],
    "city": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Delhi", "Pune"],
    "score": [85, 92, 78, 95, 88, 75],
    "age": [21, 24, 22, 27, 23, 25]
}

df = pd.DataFrame(data)
```

```python
df.groupby("city")
```

```python
df.groupby("city")["score"].mean()
```

```python
df.groupby("city")["score"].sum()
```

```python
df.groupby("city")["score"].count()
```

```python
df.groupby("city")["score"].agg(["mean", "sum", "min", "max", "count"])
```

```python
df.groupby("city").agg({
    "score": "mean",
    "age": "max"
})
```

```python
df.groupby(["city", "age"])["score"].mean()
```
